```python
import pandas as pd
import json

# 1. Load the User data (from the JSON files shown)
# Note: You should combine all user records into one list
users_data = [
    {"user_id": 1, "name": "User_1", "city": "Chennai", "membership": "Regular"},
    {"user_id": 2, "name": "User_2", "city": "Pune", "membership": "Gold"},
    # ... add all users from your JSON files here
]
df_users = pd.DataFrame(users_data)

# 2. Load the Restaurant data (from the SQL INSERT statements)
# Format: restaurant_id, name, cuisine, rating
restaurants_data = [
    [1, "Restaurant_1", "Chinese", 4.8],
    [2, "Restaurant_2", "Indian", 4.1],
    # ... add all restaurants from your SQL files here
]
df_restaurants = pd.DataFrame(restaurants_data, columns=['res_id', 'name', 'cuisine', 'rating'])

# 3. Load the Orders data (This file was not in your images, but is required)
# Assuming it has: order_id, user_id, res_id, amount, city
df_orders = pd.read_csv('orders.csv') 

# --- CALCULATIONS ---

# Merge data for comprehensive analysis
df_combined = df_orders.merge(df_users, on='user_id').merge(df_restaurants, left_on='res_id', right_on='res_id')

## 1. Total orders by Gold members
gold_orders_count = df_combined[df_combined['membership'] == 'Gold'].shape[0]

## 2. Total revenue from Hyderabad (rounded)
hyd_revenue = round(df_orders[df_orders['city'] == 'Hyderabad']['amount'].sum())

## 3. Distinct users who placed at least one order
distinct_users = df_orders['user_id'].nunique()

## 4. Average order value for Gold members (2 decimals)
aov_gold = df_combined[df_combined['membership'] == 'Gold']['amount'].mean()
aov_gold = round(aov_gold, 2)

## 5. Orders for restaurants with rating >= 4.5
high_rating_orders = df_combined[df_combined['rating'] >= 4.5].shape[0]

## 6. Orders in the top revenue city (Gold members only)
# Find top revenue city for Gold members
top_city_gold = df_combined[df_combined['membership'] == 'Gold'].groupby('city_x')['amount'].sum().idxmax()
# Count orders in that specific city for Gold members
top_city_orders = df_combined[(df_combined['membership'] == 'Gold') & (df_combined['city_x'] == top_city_gold)].shape[0]

print(f"Gold Orders: {gold_orders_count}")
print(f"Hyd Revenue: {hyd_revenue}")
print(f"Distinct Users: {distinct_users}")
print(f"Gold AOV: {aov_gold}")
print(f"Rating >= 4.5 Orders: {high_rating_orders}")
print(f"Top City (Gold) Orders: {top_city_orders}")
```
